pulling together everything this project needs: GDELT event data, sector ETF prices, risk-free rates, and FOMC dates. keeping the data collection separate from analysis so it's reproducible - anyone should be able to run this notebook and get the same raw data.

```python
import pandas as pd
import numpy as np
import yfinance as yf
import requests
import os
from datetime import datetime, timedelta
import warnings
warnings.filterwarnings('ignore')

print("imports done")
```

```python
# sector ETFs and benchmark
tickers = ['XLE', 'XLB', 'XLF', 'XLI', 'XLK', 'XLY', 'ITA', 'XLP', 'XLV', 'SPY']

# pull daily adjusted prices 2014-2023
prices = yf.download(tickers, start='2014-01-01', end='2023-12-31', auto_adjust=True)['Close']

print(prices.shape)
print(prices.head())
print(f"\nmissing values:\n{prices.isnull().sum()}")
```
